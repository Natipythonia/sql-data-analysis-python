```python
pd.read_sql(
    """
    SELECT
        Customer.FirstName,
        Customer.LastName,
        Invoice.InvoiceId,
        Invoice.InvoiceDate,
        Invoice.Total
    FROM Customer
    JOIN Invoice
        ON Customer.CustomerId = Invoice.CustomerId;
    """,
    conn
)
```

```python
pd.read_sql(
    """
    SELECT
        Customer.FirstName,
        Customer.LastName,
        SUM(Invoice.Total) AS TotalGastado
    FROM Customer
    JOIN Invoice
        ON Customer.CustomerId = Invoice.CustomerId
    GROUP BY Customer.CustomerId
    ORDER BY TotalGastado DESC;
    """,
    conn
)
```

```python
pd.read_sql(
    """
    SELECT
        Customer.FirstName,
        Customer.LastName,
        SUM(Invoice.Total) AS TotalGastado
    FROM Customer
    JOIN Invoice
        ON Customer.CustomerId = Invoice.CustomerId
    GROUP BY Customer.CustomerId
    HAVING TotalGastado > 40
    ORDER BY TotalGastado DESC;
    """,
    conn
)
```

```python
pd.read_sql(
    """
    SELECT
        EmployeeId,
        FirstName,
        LastName,
        BirthDate,
        CAST((julianday('now') - julianday(BirthDate)) / 365.25 AS INT) AS Edad
    FROM Employee;
    """,
    conn
)
```

```python
pd.read_sql(
    """
    SELECT
        Title,
        ROUND(AVG((julianday('now') - julianday(BirthDate)) / 365.25), 2) AS MediaEdad
    FROM Employee
    GROUP BY Title
    ORDER BY MediaEdad DESC;
    """,
    conn
)
```

```python
pd.read_sql("""
SELECT 
    i.InvoiceId,
    c.FirstName || ' ' || c.LastName AS CustomerName,
    i.BillingCountry,
    i.Total
FROM Invoice i
JOIN Customer c
    ON i.CustomerId = c.CustomerId
ORDER BY i.Total DESC;
""", conn)
```

```python
pd.read_sql("""
SELECT 
    BillingCountry,
    SUM(Total) AS TotalFacturado
FROM Invoice
GROUP BY BillingCountry
ORDER BY TotalFacturado DESC;
""", conn)
```
